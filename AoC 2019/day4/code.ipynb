```python
# --- PART 1 -----

# find eligible password

# Rules for passwords
#    1. it is 6 digits long
#    2. the value is between [109165, 576723]
#    3. two adjacent digits are the same (like 22 in 122345)
#    4. from digits never decrease (e.g. 11123 or 135679)

# Examples
# 111111    - YES
# 223450    - NO (decreases at 50)
# 123789    - NO (no double)

# password: the total amount of eligible passwords
```

```python
def is_eligible(num: int) -> bool:
    return(len(str(num)) == 6 and (109_165 <= num <= 576_723) and has_double_digits(num) and is_strictly_increasing(num))

def has_double_digits(num: int):
    prev_digit = 0
    for digit in str(num):
        digit = int(digit)
        if digit == prev_digit:
            return True
        prev_digit = digit
    return False

def is_strictly_increasing(num: int):
    prev_digit = 0
    for digit in str(num):
        digit = int(digit)
        if digit < prev_digit:
            return False
        prev_digit = digit
    return True


counter = 0
for i in range(109165, 576724):
    if is_eligible(i):
        counter += 1

counter
```

```python
# --- PART 2 -----

# Rule number three is modified:
#    3. there must be at least one run of exactly length 2, not 1 or 3+

# Examples
# 112233    - YES, there are several runs of length 2: 11, 22, 33
# 123444    - NO, there are no runs of length 2, only 1 and 3
# 111122    - Yes, there is a run of length 2: 22

# password: the total amount of eligible passwords
```

```python
def is_eligible(num: int) -> bool:
    return(len(str(num)) == 6 and (109_165 <= num <= 576_723) and has_double_digits(num) and is_strictly_increasing(num))

def has_double_digits(num: int) -> bool:  # noqa: F811
    run_list = []
    current_run = None
    for digit in str(num):
        if current_run and digit in current_run:
            current_run += digit
        elif current_run and digit not in current_run:
            run_list.append(current_run)
            current_run = digit
        else:
            current_run = digit

    if current_run:
        run_list.append(current_run)

    return any(len(digit) == 2 for digit in run_list)

def is_strictly_increasing(num: int):  # noqa: F811
    prev_digit = 0
    for digit in str(num):
        digit = int(digit)
        if digit < prev_digit:
            return False
        prev_digit = digit
    return True


counter = 0
for i in range(109165, 576724):
    if is_eligible(i):
        counter += 1

counter
```
