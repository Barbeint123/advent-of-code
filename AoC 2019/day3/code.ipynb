```python
# --- PART 1 -----

# The two rows in the file describes two wires
# each row has values seperated by comma (R75,D30,R83,U83, ...)
# U = up, R = right, D = down, U = up,    and then the number of steps

# The wires may intersect. Find the intersection closest to the starting point!

# password: intersection point closest to the starting point (Manhattan distance)
# NB: while the wires technically cross at the startin point, this point does not count

# Manhattan Distance
# (x1, y1) -> (x2, y2) = |x1 - x2| + |y1 - y2|
```

```python
import math


def walk_trail(command: list[str]) -> list[tuple[int, int]]:
    direction_map = {
        "R": (1, 0),
        "L": (-1, 0),
        "U": (0, 1),
        "D": (0, -1),
    }

    wire = [(0, 0)]
    for instruction in command:
        direction, steps = instruction[0], int(instruction[1:])
        dx, dy = direction_map[direction]

        for _ in range(steps):
            last_x, last_y = wire[-1]
            wire.append((last_x + dx, last_y + dy))

    return wire


def find_closest_intersection(trail_1 : list[tuple[int, int]], trail_2: list[tuple[int, int]]) -> list[tuple[int, int]]:
    intersections = []
    trail_2_set = set(trail_2)
    for coord in trail_1:
        if coord in trail_2_set:
            intersections.append(coord)

    shortest_distance = math.inf
    for coord in intersections[1:]:
        dist = manhattan_distance(coord)
        shortest_distance = min(shortest_distance, dist)

    return shortest_distance


def manhattan_distance(point_1: tuple[int, int], point_2: tuple[int, int] = (0,0)) -> int:
    x1, y1 = point_1
    x2, y2 = point_2
    return abs(x1 - x2) + abs(y1 - y2)

wire_paths = []
with open("file.txt", encoding="utf-8") as f:
    for line in f:
        line = line.strip().split(",")
        wire_paths.append(walk_trail(line))

wire_1, wire_2 = wire_paths

find_closest_intersection(wire_1,wire_2)
```

```python
# --- PART 2 -----

# Count the step to reach each intersection
# find the intersection where the sum of both wire-steps is the lowest

# if a wire visits a position on the grid multiple times,
# use the lowest steps value

# password: the fewest combined steps the wires must take to reach an intersection
```

```python
import math

def walk_trail(command: list[str]) -> list[tuple[int, int]]:
    direction_map = {
        "R": (1, 0),
        "L": (-1, 0),
        "U": (0, 1),
        "D": (0, -1),
    }

    wire = [(0, 0)]
    for instruction in command:
        direction, steps = instruction[0], int(instruction[1:])
        dx, dy = direction_map[direction]

        for _ in range(steps):
            last_x, last_y = wire[-1]
            wire.append((last_x + dx, last_y + dy))

    return wire


def find_closest_intersection(trail_1 : list[tuple[int, int]], trail_2: list[tuple[int, int]]) -> list[tuple[int, int]]:
    intersections = []
    trail_2_set = set(trail_2)
    for coord in trail_1:
        if coord in trail_2_set:
            intersections.append(coord)

    shortest_distance = math.inf
    for coord in intersections[1:]:
        dist_1 = trail_1.index(coord)
        dist_2 = trail_2.index(coord)
        shortest_distance = min(shortest_distance, dist_1 + dist_2)

    return shortest_distance


wire_paths = []
with open("file.txt", encoding="utf-8") as f:
    for line in f:
        line = line.strip().split(",")
        wire_paths.append(walk_trail(line))

wire_1, wire_2 = wire_paths

find_closest_intersection(wire_1,wire_2)
```
