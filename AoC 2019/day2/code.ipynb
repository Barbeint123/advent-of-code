```python
# --- PART 1 -----

# A csv with ints, all on a single row
# each int represents an opcode

# OPCODES
# 1 - read the next three nums, they represent indices
#     i1 + i2, stored in i3

# 2 - same as 1, but with *
#     i1 * i2, stored in i3

# 99 - halt the program


# once you are done processing an opcode, move forward 4 steps

# Ex.
# DATA
# 1,9,10,3,2,3,11,0,99,30,40,50

# step 1 (opcode 1):
#      i3 = i9 + i10        i3 = 30 + 40

# step 2 (opcode 2):
#      i0 = i3 * i11        i0 = 70 * 50


# NB:
# before running the program replace position 1 with 12
# and position 2 with 2

# Password: what is the value of index 0?
```

```python
class Intcode:
    def __init__(self, program: list[int]) -> None:
        self.pointer = 0
        self.program = program.copy()

    def execute_program(self):
        while True:
            opcode = self.program[self.pointer]
            if opcode == 1:
                self.opcode1()
            elif opcode == 2:
                self.opcode2()
            elif opcode == 99:
                break
            else:
                print("critical error")
                break
        return self.program[0]

    def opcode1(self):
        param_1 = self.program[self.pointer + 1]
        param_2 = self.program[self.pointer + 2]
        param_3 = self.program[self.pointer + 3]
        self.program[param_3] = self.program[param_1] + self.program[param_2]
        self.pointer += 4

    def opcode2(self):
        param_1 = self.program[self.pointer + 1]
        param_2 = self.program[self.pointer + 2]
        param_3 = self.program[self.pointer + 3]
        self.program[param_3] = self.program[param_1] * self.program[param_2]
        self.pointer += 4

    def __str__(self):
        return f"{self.program}"
```

```python
with open("file.txt", encoding="utf-8") as f:
    program = [int(x) for x in f.read().strip().split(",")]

program[1] = 12
program[2] = 2
computer = Intcode(program)

computer.execute_program()
print(computer.program[0])
```

```python
# --- PART 2 -----

# TERMINOLOGY
# instruction: consists of 1 opcode, 3 parameters

# once an instruction finishes, the instruction pointer increases by
# the number of values in the instruction, which is always 4 (1 op + 3 inst)

# TASK

# the value in adress 1 is called "noun"
# the value in address 2 is called "verb"
# these values are [0, 99]

# Find the noun and verb which produces the output 19690720 (in index 0)¨
# the password is 100 * noun + verb
```

```python
def find_target(blank_program, target):
    for i in range(100):
        for j in range(100):
            program = blank_program.copy()
            program[1] = i
            program[2] = j

            computer = Intcode(program)
            computer.execute_program()
            if computer.program[0] == target:
                return computer, i, j
    return None

with open("file.txt", encoding="utf-8") as f:
    blank_program = [int(x) for x in f.read().strip().split(",")]

result = find_target(blank_program, 19690720)
computer, noun, verb = result

100 * noun + verb
```
