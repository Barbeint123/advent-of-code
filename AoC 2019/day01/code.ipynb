```python
# --- PART 1 -----

# for each row:
#   divide by 3 (round down)
#   subtract 2

# Ex.
# DATA          RESULT
# 12            (12 // 3) - 2       = 2
# 14            (14 // 3) - 2       = 2
# 1969          (1969 // 3) - 2     = 654
# 100756        (100756 // 3) - 2   = 33583

# password: the sum total
# 2 + 2 + 654 + 33583 = 34241
```

```python
def calculate_fuel(num: int) -> int:
    return (num // 3) - 2


password = 0
with open("file.txt", encoding="utf-8") as f:
    for line in f:
        password += calculate_fuel(int(line))

password
```

```python
# --- PART 2 -----

# The function should be run recursively on each row
# until the step before it reaches negative
# then add all subtotals

# Ex.
# DATA          RESULT
# 12            2
# 14            2
# 1969          654 + 216 + 70 + 21 + 5     = 966
# 100756        33583 + 11192 + 3728 + ...  = 50_346

# password: the sum total
```

```python
def calculate_fuel(num: int) -> int:
    return (num // 3) - 2


def calculate_fuel_until_empty(num: int) -> int:
    subtotals = []

    while num > 0:
        num = calculate_fuel(num)
        subtotals.append(num)

    return sum(subtotals[:-1])


password = 0
with open("file.txt", encoding="utf-8") as f:
    for line in f:
        password += calculate_fuel_until_empty(int(line))

password
```
