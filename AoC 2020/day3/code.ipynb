```python
# --- PART 1 -----

# Each row consists of dots "." and hashes "#"
# Imagine the rows as repeating indefinitely in the same pattern

# Start on the top left square
# each step you move is 3 right, 1 down
# until you reach the bottom row

# password: the amount of times you land on a "#"

# Ex.   
#    DATA               STEPS                                AMOUNT OF #-s
#    ..##.......        O.##.........##.........##.....          7
#    #...#...#..        #..O#...#..#...#...#..#...#...#
#    .#....#..#.        .#....X..#..#....#..#..#....#..
#    ..#.#...#.#        ..#.#...#O#..#.#...#.#..#.#...#
#    .#...##..#.        .#...##..#..X...##..#..#...##..
#    ..#.##.....        ..#.##.......#.X#.......#.##...
#    .#.#.#....#        .#.#.#....#.#.#.#.O..#.#.#.#...
#    .#........#        .#........#.#........X.#.......
#    #.##...#...        #.##...#...#.##...#...#.X#...#.
#    #...##....#        #...##....##...##....##...#X...
#    .#..#...#.#        .#..#...#.#.#..#...#.#.#..#...X
```

```python
import math

def lands_on_hash(line: str, index: int) -> bool:
    return line[index % len(line)] == "#"


with open("file.txt") as f:
    password = 0
    position = 0

    for line in f:
        if (lands_on_hash(line.strip(), position)):
            password += 1
        position += 3
        
    print(f"password: {password}")
```

```python
# --- PART 2 -----

# Same as before, however now you need to check 5 paths:
#       right 1, down 1
#       right 3, down 1
#       right 5, down 1
#       right 7, down 1
#       right 1, down 2


# Ex. 
#   Right 1, down 1             right 5, down 1                                                      right 1, down 2                       
#   O.##.......                 O.##.........##.........##.........##.........##.......              O.##.......
#   #O..#...#..                 #...#O..#..#...#...#..#...#...#..#...#...#..#...#...#..              #...#...#..
#   .#O...#..#.                 .#....#..#O.#....#..#..#....#..#..#....#..#..#....#..#.              .X....#..#.
#   ..#O#...#.#                 ..#.#...#.#..#.X...#.#..#.#...#.#..#.#...#.#..#.#...#.#              ..#.#...#.#
#   .#..O##..#.                 .#...##..#..#...##..X..#...##..#..#...##..#..#...##..#.              .#O..##..#.
#   ..#.#X.....                 ..#.##.......#.##.......#O##.......#.##.......#.##.....              ..#.##.....
#   .#.#.#O...#                 .#.#.#....#.#.#.#....#.#.#.#..O.#.#.#.#....#.#.#.#....#              .#.X.#....#
#   .#.....O..#                 .#........#.#........#.#........#.#O.......#.#........#              .#........#
#   #.##...#O..                 #.##...#...#.##...#...#.##...#...#.##...X...#.##...#...              #.##O..#...
#   #...##...O#                 #...##....##...##....##...##....##...##....##O..##....#              #...##....#
#   .#..#...#.X                 .#..#...#.#.#..#...#.#.#..#...#.#.#..#...#.#.#..#.O.#.#              .#..#O..#.#

# Password: path counts multiplied
#       right 1, down 1     - 2 
#       right 3, down 1     - 7
#       right 5, down 1     - 3
#       right 7, down 1     - 4
#       right 1, down 2     - 2
# 2*7*3*4*2 = 336 
```

```python
import math

def lands_on_hash(line: str, index: int) -> bool:
    return line[index % len(line)] == "#"

def find_hashes_in_trail(all_rows: list[str], right: int, down: int) -> int:
    position = 0
    hash_count = 0

    for index, row in enumerate(all_rows):
        if index % down != 0:
            continue
    
        if lands_on_hash(row, position):
            hash_count += 1
        position += right

    return hash_count

with open("file.txt") as f:
    rows = [line.strip() for line in f]
    paths = [(1,1), (3,1), (5,1), (7,1), (1,2)]

    password = 1
    for right, down in paths:
        password *= find_hashes_in_trail(rows, right, down)

    print(password)
```
