```python
# --- PART 1 -----

# Find the two integers that sum to 2020
# then multiply the integers to obtain the password

# Ex.   
# DATA          RELEVANT INTS   
# 1721          1721 + 299 = 2020
# 979            
# 366          
# 299        
# 675
# 1456          

# password: the product (multiplied)
# 1721 * 299 = 514579
```

```python
from itertools import combinations

def find_ints(arr: list[int]) -> tuple[int, int] | None:
    for a, b in combinations(arr, 2):
        if a + b == 2020:
            return a, b

with open("file.txt") as f:
    numbers = [int(line) for line in f]
    a, b = find_ints(numbers)
    password = a * b
    print(password)
```

```python
# --- PART 2 -----

# Find the THREE integers that sum to 2020 
# 979 + 366 + 675 = 2020

# password: the product (multiplied)
# 979 * 366 * 675 = 241861950
```

```python
from itertools import combinations

def find_ints2(arr: list[int]) -> tuple[int, int] | None:
    for a, b, c in combinations(arr, 3):
        if a + b + c == 2020:
            return a, b, c

with open("file.txt") as f:
    numbers = [int(line) for line in f]
    a, b, c = find_ints2(numbers)
    password = a * b * c
    print(password)
```
