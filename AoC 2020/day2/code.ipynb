```python
# --- PART 1 -----

# Each row in the data consists of:
#   an interval, ex 1-3: a    -     the amount of times "a" is allowed in the passowrd
#   the password, ex abcde

# find out how many passwords are valid

# Ex.   
# DATA                      VALID?
# 1-3 a: abcde              Y              
# 1-3 b: cdefg              N
# 2-9 c: ccccccccc          Y

# password: the number of valid passwords
```

```python
import re

def parse_line(line: str) -> tuple[int, int, str, str]:
    minimum, maximum, char, password = re.split(r"[-: ]+", line) 
    return(int(minimum), int(maximum), char, password)

def is_valid(parsed_line: tuple[int, int, str, str]) -> bool:
    minimum, maximum, char, password = parsed_line
    amount = password.count(char)
    return(minimum <= amount <= maximum )


with open("file.txt") as f:
    lines = [is_valid(parse_line(str(line).strip())) for line in f]
    print(lines.count(True))
```

```python
# --- PART 2 -----

# now 1-3 means position 1 OR 3. 
# the char must by present at 1 XOR 3  (i.e. exactly one place)
# NB: 1-indexed

# Ex.   
# DATA                      VALID?
# 1-3 a: abcde              Y, pos1 contains "a", while pos3 does not            
# 1-3 b: cdefg              N, neither pos1 nor pos3 contains "b"
# 2-9 c: ccccccccc          Y, both pos2 and pos9 contains "b"

# password: the number of valid passwords
```

```python
import re

def parse_line(line: str) -> tuple[int, int, str, str]:
    pos1, pos2, char, password = re.split(r"[-: ]+", line) 
    return(int(pos1), int(pos2), char, password)

def is_valid2(parsed_line: tuple[int, int, str, str]) -> bool:
    pos1, pos2, char, password = parsed_line
    ch1, ch2 = password[pos1-1], password[pos2-1]
    return (ch1 == char) != (ch2 == char)


with open("file.txt") as f:
    lines = [is_valid2(parse_line(str(line).strip())) for line in f]
    print(lines.count(True))
```
