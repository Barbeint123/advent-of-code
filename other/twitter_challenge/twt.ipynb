```python
# Convert a non-negative integer num to its English words representation (leetcode 273)

# E X A M P L E S
# Input: num = 123
# Output: "One Hundred Twenty Three"

# EX 2
# Input: num = 12345
# Output: "Twelve Thousand Three Hundred Forty Five"

# EX 3
# Input: num = 1234567
# Output: "One Million Two Hundred Thirty Four Thousand Five Hundred Sixty Seven"
```

```python
import re

below_20 = [
    "", "One", "Two", "Three", "Four", "Five",
    "Six", "Seven", "Eight", "Nine", "Ten", "Eleven", "Twelve",
    "Thirteen", "Fourteen", "Fifteen", "Sixteen",
    "Seventeen", "Eighteen", "Nineteen"
]
tens = [
    "", "", "Twenty", "Thirty", "Forty",
    "Fifty", "Sixty", "Seventy", "Eighty", "Ninety"
]

def numberToString(n, s=""):
  # "0" edgecase
  if n == 0 and s =="":
    return "Zero"

  # Base case
  if n is None:
    string_with_spaces = re.sub(r'(?<!^)(?=[A-Z])', ' ', s) 
    return string_with_spaces

  # Recursive cases
  if n < 20:
    s+= f"{below_20[n]}"
    return numberToString(None, s)
  
  elif n < 100:
    tenner, rest = divmod(n, 10)
    s += f"{tens[tenner]}"
    return numberToString(rest, s)
  
  elif n < 1000:
    hundreds, rest = divmod(n, 100)
    s += f"{below_20[hundreds]}Hundred"
    return numberToString(rest, s)
  
  elif n < 1_000_000:
    thousands, rest = divmod(n, 1_000)
    s += f"{numberToString(thousands)}Thousand"
    return numberToString(rest, s)
  
  elif n < 1_000_000_000:
    millions, rest = divmod(n, 1_000_000)
    s += f"{numberToString(millions)}Million"
    return numberToString(rest, s)
  
  else:
    billions, rest = divmod(n, 1_000_000_000)
    s += f"{numberToString(billions)}Billion"
    return numberToString(rest, s)
  
numberToString(100)
```
