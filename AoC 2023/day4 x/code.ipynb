```python
# --- PART 1 -----

# Each row has its ID number (e.g. Card 2:)
# followed by a two lists of numbers seperated by an "|"

# List 1 - winning nums 
# List 2 - nums you have

# the first winning num is worth 1 point
# each winning num after that doubles the points you have


# EX:                                                       Matching               Points
# Card 1: 41 48 83 86 17 | 83 86  6 31 17  9 48 53          48, 83, 17, 86         1*2*2*2 = 8


# password: total sum of points
```

```python
password = 0

with open("file.txt", encoding="utf-8") as f:
    for line in f:
        _, nums = line.split(":")
        scoring_nums, our_nums = (el.split() for el in nums.split("|"))
        
        score = 0
        for num in scoring_nums:
            if num in our_nums:
                score = 1 if score == 0 else score * 2
        password += score

password 
```
