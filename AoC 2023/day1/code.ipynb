```python
# --- PART 1 -----

# find the first and last digits of each line

# Ex. 
# 1abc2  -->  12
# pqr3stu8vwx  -->  38
# a1b2c3d4e5f  -->  15
# treb7uchet  -->  77

# then get the sum total: 12 + 38 + 15 + 77 = 142
```

```python
def find_digits(s: str) -> int:
    first = None
    last = None 

    for char in s:
        if char.isdigit(): 
            first = char
            break

    for char in reversed(s):
        if char.isdigit(): 
            last = char
            break

    return int(first + last)


password = 0
with open("test.txt", encoding="utf-8") as f:
    for line in f:
        password += find_digits(line)

print(password)
```

```python
# --- PART 2 -----

# find the first and last digits of each line
# # NB: digits may be written as strings as well

# Ex. 
# two1nine --> 29
# eightwothree --> 83
# abcone2threexyz --> 13
# xtwone3four --> 24
# 4nineeightseven2 --> 42
# zoneight234 --> 14
# 7pqrstsixteen --> 76

# then get the sum
```

```python
number_dict = {
    "one": 1, "two": 2, "three": 3,
    "four": 4, "five": 5, "six": 6,
    "seven": 7, "eight": 8, "nine": 9
}


def convert_alphanum_to_num(s: str) -> int:
    new_string = s
    for num in number_dict.keys():
        new_string = new_string.replace(num, num[:1] + str(number_dict[num]) + num[-1:])
    return new_string


def find_digits(s: str) -> int:
    first = None
    last = None 

    for char in s:
        if char.isdigit(): 
            first = char
            break

    for char in reversed(s):
        if char.isdigit(): 
            last = char
            break

    return int(first + last)


password = 0
with open("file.txt", encoding="utf-8") as f:
    for line in f:
        num = convert_alphanum_to_num(line)
        password += find_digits(num)

password
```
