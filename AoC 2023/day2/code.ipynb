```python
# --- PART 1 -----

# Each row has its ID number (e.g. Game 2:)
# followed by a semicolon-separated rgb values (e.g. 3 blue, 4 red;)

# We have a limit of:
# 12 red, 13 green, 14 blue

# each singular comma-seperated-value can never surpass these limits
# for instance, 14 green is ILLEGAL (max 13 green) 

# EX:
# LINES                                                                        POSSIBLE?
# Game 1: 3 blue, 4 red; 1 red, 2 green, 6 blue; 2 green                       YES
# Game 2: 1 blue, 2 green; 3 green, 4 blue, 1 red; 1 green, 1 blue             YES
# Game 3: 8 green, 6 blue, 20 red; 5 blue, 4 red, 13 green; 5 green, 1 red     NO
# Game 4: 1 green, 3 red, 6 blue; 3 green, 6 red; 3 green, 15 blue, 14 red     NO
# Game 5: 6 red, 1 blue, 3 green; 2 blue, 1 red, 2 green                       YES

# password: add all valid IDs together  (1 + 2 + 5 = 8)
```

```python
import re

color_map = {
    "red": 12,
    "green": 13,
    "blue": 14,
}

password = 0

with open("file.txt", encoding="utf-8") as f:
    for line in f:
        game, data = line.rstrip().split(":")
        game_id = int(game.split()[1])

        is_safe = True

        for part in re.split(r"[;,]", data):
            amount, color = part.strip().split()

            if int(amount) > color_map[color]:
                is_safe = False
                break

        if is_safe:
            password += game_id
```

```python
# --- PART 2 -----

# in each game you played, what is the fewest number of cubes of each color that could have been in the bag to make the game possible?
# i.e. the maximum for each color

# EX
# LINES                                                                                                     SUM
# Game 1: 3 blue, 4 red; 1 red, 2 green, 6 blue; 2 green --> 4 red, 2 green, 6 blue                         4*2*6 = 48
# Game 2: 1 blue, 2 green; 3 green, 4 blue, 1 red; 1 green, 1 blue --> 1 red, 3 green, 4 blue               1*3*4 = 12
# Game 3: 8 green, 6 blue, 20 red; 5 blue, 4 red, 13 green; 5 green, 1 red --> 20 red, 13 green, 6 blue     20*13*6 = 1560
# Game 4: 1 green, 3 red, 6 blue; 3 green, 6 red; 3 green, 15 blue, 14 red --> 14 red, 3 green, 15 blue     14*3*15 = 630
# Game 5: 6 red, 1 blue, 3 green; 2 blue, 1 red, 2 green --> 6 red, 3 green, 2 blue                         6*3*2 = 36

# 48+12+1560+630+36 = 2286
```

```python
import re


password = 0

with open("file.txt", encoding="utf-8") as f:
    for line in f:
        data = line.rstrip().split(":")[1]

        color_map = {
            "red": 0,
            "green": 0,
            "blue": 0,
        }

        for part in re.split(r"[;,]", data):
            amount, color = part.strip().split()

            amount = int(amount)

            if color_map[color] < amount:
                color_map[color] = amount
            
        line_sum = 1
        for amount in color_map.values():
            line_sum *= amount
        password += line_sum

password
```
