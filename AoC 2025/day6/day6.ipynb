```python
# --- PART 1 -----

# The data represents a vertical list of algebraic expressions
# there are 5 rows; the first 4 are integers, the last is operators (+, *)
# the numbers / operators in each row are seperated by white space
# The left/right alignment of numbers within each problem can be ignored.

# EX:
# DATA                      INTERPRETATION
# 123 328  51 64            123 * 45 * 6 = 33210
#  45 64  387 23            328 + 64 + 98 = 490
#   6 98  215 314           51 * 387 * 215 = 4243455
# *   +   *   +             64 + 23 + 314 = 401

# password: 33210 + 490 + 4243455 + 401 = 4277556
```

```python
import numpy as np

with open("day6.txt", encoding="utf-8") as f:
    data = [line.split() for line in f]

horizontal_arr = np.array(data).T

password = 0
for expr in horizontal_arr:
    nums = list(map(int, expr[:4]))
    operator = expr[4]

    result = np.prod(nums) if operator == "*" else sum(nums)
    password += result

password
```

```python
# --- PART 2 -----

# the expressions are read from right-to-left in columns
# Each number is given in its own column,
# NB: nums may either be right- or left aligned
# NB: numbers are between 1 and 4 digits long

# EX:
# DATA                      INTERPRETATION
# 123 328  51 64            4 + 431 + 623 = 1058
#  45 64  387 23            175 * 581 * 32 = 3253600
#   6 98  215 314           8 + 248 + 369 = 625
# *   +   *   +             356 * 24 * 1 = 8544

# password: 1058 + 3253600 + 625 + 8544 = 3263827
```

```python
import numpy as np


def largest_digit_in_array(arr):
    return max(len(str(line)) for line in arr)


def calculate_column(arr):
    operator = arr[-1]
    nums = []
    num = ""
    for i in range(0, len(str(arr[0]))):
        for j in range(0, len(arr) - 1):
            current_digit = arr[j][-i - 1]
            if current_digit != " ":
                num += current_digit
        nums.append(int(num))
        num = ""
    return sum(nums) if operator == "+" else np.prod(nums)


with open("day6.txt", encoding="utf-8") as f:
    data = [line.rstrip("\n") for line in f]

horizontal_arr = np.array([line.split() for line in data]).T
largest_digit_per_column = [largest_digit_in_array(row) for row in horizontal_arr]

aligned_arr = []

for line in data:
    start = 0
    temporary_arr = []

    for column_size in largest_digit_per_column:
        temporary_arr.append(line[start : start + column_size])
        start += column_size + 1

    aligned_arr.append(temporary_arr)

# remove spacing from operators
aligned_arr[-1] = [op.strip() for op in aligned_arr[-1]]

horizontal_aligned_arr = np.array(aligned_arr).T

password = sum([calculate_column(line) for line in horizontal_aligned_arr])
password
```
