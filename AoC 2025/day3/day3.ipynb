```python
# --- PART 1 -----

# find the two digits (batteries) in each row (bank)
# which combines to create the highest two-digit number (largest joltage)
# NB: must preserve its order in relation to each other

# password: add all two-digit numbers together

# EX:
# NUMBERS                 DIGITS
# 987654321111111         98
# 811111111111119         89
# 234234234234278         78
# 818181911112111         92
```

```python
def find_largest_joltage(bank):
    length = len(bank)

    tens = int(bank[0])
    ones = -1

    for i in range(1, length):
        current = int(bank[i])

        if current > tens and i != length - 1:
            tens = current
            ones = -1

        elif current > ones:
            ones = current

    return tens * 10 + ones


total = 0
with open("day3.txt", newline=None, encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        total += find_largest_joltage(line)

total
```

```python
# --- PART 2 -----
# same as before, however now we are looking for the largest 12-digit number

# EX:
# NUMBERS                 DIGITS
# 987654321111111         987654321111
# 811111111111119         811111111119
# 234234234234278         434234234278
# 818181911112111         888911112111
```

```python
def find_largest_digit(bank, digits_left):
    digits = "9876543210"

    for digit in digits:
        if digit in bank:
            index = bank.find(digit)
            if index < (len(bank) - digits_left + 1):
                largest_digit = digit
                break

    return largest_digit, bank[index:]


def find_largest_joltage(bank):
    largest_joltage = ""

    for i in range(12, 0, -1):
        largest_digit, new_bank = find_largest_digit(bank, i)
        largest_joltage += largest_digit
        bank = new_bank[1:]

    return int(largest_joltage)


with open("day3.txt", encoding="utf-8", newline=None) as f:
    total = 0
    for line in f:
        line = line.strip()
        total += find_largest_joltage(line)
    print(total)
```
