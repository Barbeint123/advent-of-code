```python
# --- PART 1 -----

# each row in the data represents the coordinates of a node (x,y,z)

# iterate through the data, connecting the closest two nodes for each iteration
# connected nodes form a cluster
# continue until all nodes have at least one connection

# NB: closest according to euclidian distance

# password: the three largest clusters multiplied
```

```python
import math

def euclidean_distance(p1, p2):
    return math.dist(p1, p2)  # built-in, cleaner


with open("day8_test.txt", encoding="utf-8") as f:
    data = [list(map(int, line.strip().split(","))) for line in f]


closest = {}

for i, p1 in enumerate(data):
    closest_dist = math.inf
    closest_idx = None

    for j, p2 in enumerate(data):
        if i == j:
            continue

        current_dist = euclidean_distance(p1, p2)
        if current_dist < closest_dist:
            closest_dist = current_dist
            closest_idx = j

    closest[i] = (closest_dist, closest_idx)

sorted_closest = sorted(closest.items(), key=lambda x: x[1][0])
sorted_closest
```

```python
import numpy as np
import math
from collections import defaultdict


def euclidean_distance(p1, p2):
    return math.dist(p1, p2)


with open("day8_test.txt", encoding="utf-8") as f:
    data = [list(map(int, line.strip().split(","))) for line in f]


# finding all shortest distances
d = defaultdict(lambda: [0,0])

closest_distance = math.inf
closest_coord = None

for p1_index, p1_coord in enumerate(data):
    for p2_index, p2_coord in enumerate(data):

        if p1_index == p2_index:
            continue

        current_distance = euclidean_distance(p1_coord, p2_coord)
        if current_distance < closest_distance:
            closest_distance = current_distance
            closest_coord = p2_index

    if d[closest_coord][1] != p1_index:
        d[p1_index] = [closest_distance, closest_coord]
    closest_distance = math.inf

for i in range(0, 20):
    if d[i] == [0, 0]:
        del d[i]

sorted_closest = sorted(closest.items(), key=lambda x: x[1][0])
sorted_closest
```

```python
import numpy as np
import pandas as pd
import math


def euclidean_distance(x, y):
    p1 = np.array([x])
    p2 = np.array([y])
    return np.linalg.norm(p2 - p1)


with open("day8_test.txt", encoding="utf-8") as f:
    data = [list(map(int, line.strip().split(","))) for line in f]


# finding all shortest distances
d = []

closest_distance = math.inf
closest_coord = None

for p1_index, p1_coord in enumerate(data):
    for p2_index, p2_coord in enumerate(data):
        # print(p1_index, p2_index)
        if p1_index == p2_index:
            continue

        current_distance = euclidean_distance(p1_coord, p2_coord)
        # print(current_distance)
        if current_distance < closest_distance:
            closest_distance = current_distance
            closest_coord = p2_index

    d.append(
        {
          "from": p1_coord,
          "to": data[closest_coord],
          "distance": closest_distance,
         }
    )

    closest_distance = math.inf


pd.DataFrame(d)
```
