```python
# --- PART 1 -----

# The data represents points in a 2d matrix
# each row is one point (e.g. 7,1)
# from two points one can create a rectangle, with each point being opposite corners

# NB: they calculate area strangely:
# (width+1) * (length+1)

# EX
# Starting matrix     Example rectangle (24)     Largest rectangle (50)
# ..............      ..............              ..............
# .......#...#..      .......#...#..              ..OOOOOOOOO#..
# ..............      ..............              ..OOOOOOOOOO..
# ..#....#......      ..#....#......              ..OOOOOOOOOO..
# ..............      ..............              ..OOOOOOOOOO..
# ..#......#....      ..#OOOOOOOOO..              ..#OOOOOOOOO..
# ..............      ..OOOOOOOOOO..              ..............
# .........#.#..      ..OOOOOOOOO#..              .........#.#..
# ..............      ..............              ..............

# password: the area of the largest rectangle
```

```python
from itertools import combinations

def get_volume(p1, p2):
    return (abs(p1[0] - p2[0])+1) * (abs(p1[1] - p2[1])+1)


points = [tuple(map(int, line.split(","))) for line in open("day9.txt").read().splitlines()]

largest_area = -1
for p1, p2 in combinations(points, 2):
    current_area = get_volume(p1, p2)
    if current_area > largest_area:
        largest_area = current_area

largest_area
```
