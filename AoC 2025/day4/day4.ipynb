```python
# --- PART 1 -----

# there is a matrix consisting of paperrolls (@) and open spaces (.)
# paperrolls may be bulldozed if they have fewer than 4 adjacent paperrolls
# password: number of paperrolls which may be bulldozed

# EX:
#
#   ..@@.@@@@.    ..XX.XX@X.
#   @@@.@.@.@@    X@@.@.@.@@
#   @@@@@.@.@@    @@@@@.X.@@
#   @.@@@@..@.    X.@@@X..X.
#
#   Password = 10
```

```python
import numpy as np
from scipy.ndimage import convolve

with open("day4.txt", encoding="utf-8") as f:
    lines = f.read().splitlines()

target_present = np.array([list(line) for line in lines]) == "@"
target_grid = target_present.astype(int)

kernel = np.array([[1, 1, 1], [1, 0, 1], [1, 1, 1]])

neighbor_counts = convolve(target_present.astype(int), kernel, mode="constant", cval=0)

np.sum(target_present & (neighbor_counts < 4))
```

```python
# --- PART 2 -----

# the paperrolls which can be bulldozed are removed

# this process of removing paperrolls may continue as
# long as paperrolls are removed each iteration
```

```python
import numpy as np
from scipy.ndimage import convolve

password = 0

with open("day4.txt", encoding="utf-8") as f:
    lines = f.read().splitlines()

kernel = np.array([[1, 1, 1], [1, 0, 1], [1, 1, 1]])

target_present = np.array([list(line) for line in lines]) == "@"
target_grid = target_present.astype(int)

while True:
    neighbor_counts = convolve(target_grid, kernel, mode="constant", cval=0)

    removable_targets = (target_grid == 1) & (neighbor_counts < 4)
    total_removable = np.sum(removable_targets)

    if total_removable > 0:
        target_grid[removable_targets] = 0
        password += total_removable
    else:
        break

password
```
