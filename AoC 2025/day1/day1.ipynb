```python
# --- PART 1 -----

# pointer starts at 50
# and has number [0, 99]

# day1.txt has a list of commands <direction><steps>
# R8 means increase the pointer by 8,
# L23 means decrase the pointer by 23

# 11 -R8-> 19
# 19 -L19-> 0
# 5 -L10-> 95

# password = number of times pointing at 0
```

```python
password = 0
pointer = 50

with open("day1.txt", encoding="utf-8") as f:
    for line in f:
        sign = 1 if line[0] == "R" else -1
        steps = int(line[1:]) * sign  
        
        pointer = (pointer + steps) % 100

        if pointer == 0:
            password += 1

password
```

```python
# --- PART 2 -----

# the password also += 1, whenever the pointer ticks past 0

# 50 -L68-> 82,   password += 1
# 10 -L200-> 10,  password += 2
# 52 -R48-> 0,    password +=1
```

```python
pointer = 50
password = 0

with open("day1.txt", encoding="utf-8") as f:
    for line in f:
        direction = line[0]
        steps = int(line[1:])

        if direction == "R":
            password += (pointer + steps) // 100
            pointer = (pointer + steps) % 100

        if direction == "L":
            if pointer == steps % 100:
                password += 1
            if pointer == 0:
                password -= 1
            password += abs((pointer - steps) // 100)
            pointer = (pointer - steps) % 100
        # print(f"pointer: {pointer} password: {password}")

password
```
