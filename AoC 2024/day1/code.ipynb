```python
# --- PART 1 -----

# two numbers for each row, separated by 3 spaces

# find the difference between the smallest number in
# coumn one and column two

# then the difference between the second smallest number
# in column one and column two
# (and so on)

# password: the sum of the differences

# EX:
# ORIGINAL          SORTED         DIFF
# 3   4             1   3           2
# 4   3             2   3           1
# 2   5             3   3           0
# 1   3             3   4           1
# 3   9             3   5           2
# 3   3             4   9           5

# Password: 2+1+1+2+5 = 11
```

```python
password = 0

left = []
right = []

with open("file.txt", encoding="utf-8") as f:
    for line in f:
        l, r = map(int, line.split())
        left.append(l)
        right.append(r)
    
    
left = sorted(left)
right = sorted(right)

for l, r in zip(left, right):
    password += abs(l - r)

password
```

```python
# --- PART 2 -----

# find how often number appear in both lists
# find similarity scores:
# num_in_col1 * times_in_col_2 

# password: the sum of the similarity scores

# EX:
# ORIGINAL       C1  AMOUNT_IN_C2     SIM SCORE
# 3   4          3        3           3*3 = 9
# 4   3          4        1           4*1 = 4
# 2   5          2        0           2*0 = 0
# 1   3          1        0           1*0 = 0
# 3   9          3        3           3*3 = 9
# 3   3          3        3           3*3 = 9

# Password: 9+4+9+9 = 31
```

```python
from collections import Counter

password = 0

left = []
right = []

with open("file.txt", encoding="utf-8") as f:
    for line in f:
        l, r = map(int, line.split())
        left.append(l)
        right.append(r)
    
    
left = sorted(left)
right_counts = Counter(right)


for num in left:
    password += num * right_counts[num]

password
```
