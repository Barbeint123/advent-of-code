```python
# --- PART 1 -----

# each row contains a sequence of numbers
# the sequences should:
#  1. be strictly increasing / decreasing
#  2. only increase / decrease in steps of [1, 3]

# password: number of adherent rows

# EX:
# DATA              ADHERENT? 
# 7 6 4 2 1         Yes
# 1 2 7 8 9         No, 2 7 is an increase of 5
# 9 7 6 2 1         No, 6 2 is a decrease of 4
# 1 3 2 4 5         No, 1 3 2 is not strictly increasing / decreasing
# 8 6 4 4 1         No, 4 4 is not strictly increasing / decreasing
# 1 3 6 7 9         Yes

# Password: 2
```

```python
password = 0

def is_valid_step(x, y):
    return abs(x-y) in [1, 2, 3]

def is_increasing(x, y):
    return x < y


with open("file.txt", encoding="utf-8") as f:
    for line in f:
        row = list(map(int, line.split()))
    
        direction = is_increasing(row[0], row[1])

        passed = True
        for i in range(len(row) - 1):
            adherent = is_valid_step(row[i], row[i+1])
            current_direction = is_increasing(row[i], row[i+1])

            if not adherent or direction != current_direction:
                passed = False
                break

        if passed:
            password += 1

password
```

```python
password = 0

def is_valid_step(x, y):
    return abs(x-y) in [1, 2, 3]

def is_increasing(x, y):
    return x < y


with open("file.txt", encoding="utf-8") as f:
    for line in f:
        row = list(map(int, line.split()))
    
        pairs = list(zip(row, row[1:]))
        direction = is_increasing(*pairs[0])

        if all(is_valid_step(a, b) and is_increasing(a, b) == direction for a, b in pairs):
            password += 1

password
```
